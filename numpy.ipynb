```python
j = np.array([[1,2,5],[2,3,7],[3,4,8],[5,6,9]])
j
"""
🔷 Beginner Rule (memorize this)
Use single index (0) → flatten result
Use slice (0:1) → keep 2D structure

:      → all elements
0:     → from 0 to end
:2     → from start to index 1
"""
```

```python
print(j[2,1:]) #q1
print(j[3,0:2]) #q2
print(j[0,-1]) ##q3
print(j[3,0]) #q3 other answer
print(j[2,0:2],j[3,0:2]) #q4
print(j[2:4, 0:2]) #q4 other answer
"""
2:4 → selects rows 2 and 3
0:2 → selects columns 0 and 1
"""
print(j[0:,0])
print(j[0:,0:1])
```

```python
#reshape
"""
reshape() changes the shape (dimensions) of an array without changing its data.
array.reshape(rows, columns)
Or more generally:
array.reshape(new_shape)
rows × columns = total elements, otherwise error
"""

jk = np.array([[1,2,3,4],[4,5,6,7],[9,8,6,5]])
jk
```

```python
jk.reshape([6,2])
```

```python
jk.reshape([4,3])
```

```python
# q1a)create one dimensional array multiple of 4 upto 40 (arange)
ar = np.arange(4,41,4)
ar #[ 4  8 12 16 20 24 28 32 36 40]

# q1b)convert it in possible 2 dimensional shapes

"""The array has 10 elements, so valid 2D shapes must multiply to 10:

(1 × 10)
(2 × 5)
(5 × 2)
(10 × 1)"""

# 1 x 10
print(ar.reshape(1, 10))

# 2 x 5
print(ar.reshape(2, 5))

# 5 x 2
print(ar.reshape(5, 2))

# 10 x 1
print(ar.reshape(10, 1))
```
