```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test =  train_test_split(X,Y,test_size=0.2,random_state=4)
```

```python
X_train.head()
```

```python
X_train.shape
```

```python
Y_train.shape
```

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler

scaler = StandardScaler()

X_train = pd.DataFrame(scaler.fit_transform(X_train))
X_test = pd.DataFrame(scaler.transform(X_test))
```

```python
# KNN

from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=9) # any odd number
knn.fit(X_train,Y_train)
```

```python
Y_pred = knn.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
print("Accuracy score: ",accuracy_score(Y_pred,Y_test)*100)
print("Precision score: ",precision_score(Y_pred,Y_test,average="weighted")*100)
# weighted is used when the Y is not binary classification (not 0,1)
# Compute the metric for each class separately, then average them using each class’s support (number of true samples) as the weight.
print("Recall score: ",recall_score(Y_pred,Y_test,average="weighted")*100)
print("F1 score: ",f1_score(Y_pred,Y_test,average="weighted")*100)

```
