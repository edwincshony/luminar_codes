```python
df.drop_duplicates(keep="first",inplace=True)
```

```python
X = df.iloc[:,:-1]
Y = df.iloc[:,-1]
```

```python
X.head()
```

```python
Y.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
X["GENDER"] = le.fit_transform(X["GENDER"])
```

```python
X.head()
```

```python
le = LabelEncoder()
Y = pd.DataFrame(le.fit_transform(Y))
```

```python
Y.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test =  train_test_split(X,Y,test_size=0.2,random_state=4) #Same rows go into train/test every time.
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
df.shape
```

```python
X_train.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_train = pd.DataFrame(scaler.fit_transform(X_train))
X_test = pd.DataFrame(scaler.transform(X_test))

# fit_transform -> training data
# transform     -> test data

# fit_transform()
# Does 2 things together:
# Learns the min and max values from data (fit)
# Scales the data (transform)

# transform()
# Only scales using already learned min/max values.
```

```python
# KNN

from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=3) # any odd number
knn.fit(X_train,Y_train)
```

```python
Y_pred = knn.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
print("===== KNN =====")

print("Accuracy score",accuracy_score(Y_pred,Y_test)*100)
print("Precision score",precision_score(Y_pred,Y_test)*100)
print("Recall score",recall_score(Y_pred,Y_test)*100)
print("F1 score",f1_score(Y_pred,Y_test)*100)
```

```python
from sklearn.naive_bayes import GaussianNB
nb = GaussianNB()
nb.fit(X_train,Y_train)
Y_predict = nb.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score

print("===== naive bayes =====")


print("accuracy score",accuracy_score(Y_predict,Y_test)*100)
print("precision_score",precision_score(Y_predict,Y_test)*100)
print("recall_score",recall_score(Y_predict,Y_test)*100)
print("f1_score",f1_score(Y_predict,Y_test)*100)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(criterion="gini",max_depth=4)
dt.fit(X_train,Y_train)
Y_predict1 = dt.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score

print("===== decision tree =====")


print("accuracy score",accuracy_score(Y_predict1,Y_test)*100)
print("precision_score",precision_score(Y_predict1,Y_test)*100)
print("recall_score",recall_score(Y_predict1,Y_test)*100)
print("f1_score",f1_score(Y_predict1,Y_test)*100)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(X_train,Y_train)
Y_predict2 = rf.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score

print("===== Random Forest =====")

print("accuracy score",accuracy_score(Y_predict2,Y_test)*100)
print("precision_score",precision_score(Y_predict2,Y_test)*100)
print("recall_score",recall_score(Y_predict2,Y_test)*100)
print("f1_score",f1_score(Y_predict2,Y_test)*100)
```
