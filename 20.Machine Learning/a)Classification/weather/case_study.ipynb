```python
X = df.iloc[:,:-1]
Y = df.iloc[:,-1]
```

```python
X.head()
```

```python
X["date"] = pd.to_datetime(X["date"])
```

```python
X["year"] = X["date"].dt.year
X["month"] = X["date"].dt.month
X["day"] = X["date"].dt.day
```

```python
X.head()
```

```python
X.drop(columns=["date"],axis=0,inplace=True)
```

```python
X.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
Y=pd.DataFrame(le.fit_transform(Y))
```

```python
Y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
Y.value_counts()
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier()
knn.fit(X_train,Y_train)
Y_pred = knn.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score

print("===== KNN =====")

print("accuracy score",accuracy_score(Y_pred,Y_test)*100)
print("precision_score",precision_score(Y_pred,Y_test,average="weighted")*100)
print("recall_score",recall_score(Y_pred,Y_test,average="weighted")*100)
print("f1_score",f1_score(Y_pred,Y_test,average="weighted")*100)
```

```python
from sklearn.naive_bayes import GaussianNB
nb = GaussianNB()
nb.fit(X_train,Y_train)
Y_predict = nb.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score

print("===== naive bayes =====")


print("accuracy score",accuracy_score(Y_predict,Y_test)*100)
print("precision_score",precision_score(Y_predict,Y_test,average="weighted")*100)
print("recall_score",recall_score(Y_predict,Y_test,average="weighted")*100)
print("f1_score",f1_score(Y_predict,Y_test,average="weighted")*100)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(criterion="entropy",max_depth=4)
dt.fit(X_train,Y_train)
Y_predict1 = dt.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score

print("===== decision tree =====")

print("accuracy score",accuracy_score(Y_predict1,Y_test)*100)
print("precision_score",precision_score(Y_predict1,Y_test,average="weighted")*100)
print("recall_score",recall_score(Y_predict1,Y_test,average="weighted")*100)
print("f1_score",f1_score(Y_predict1,Y_test,average="weighted")*100)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(X_train,Y_train)
Y_predict2 = rf.predict(X_test)
```
