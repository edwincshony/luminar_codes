```python
X = df[["Annual Income (k$)","Spending Score (1-100)"]]
```

```python
X.head()
```

```python
wcss = []
from sklearn.cluster import KMeans
```

```python
for i in range(1,11):

    km = KMeans(n_clusters=i,random_state=2)

    km.fit(X)

    print(f"K = {i},WCSS = {km.inertia_}")

    wcss.append(km.inertia_)
```

```python
plt.plot(range(1,11),wcss)
plt.show()
```

```python
km = KMeans(n_clusters=5,random_state=2)
km.fit(X)
Y_pred = km.predict(X)
```

```python
print(Y_pred)
```

```python
X["Cluster"] = Y_pred
```

```python
X.head()
```

```python
plt.scatter(x=X["Spending Score (1-100)"],y=X["Annual Income (k$)"],c=X["Cluster"])
plt.xlabel("Spending Score")
plt.ylabel("Annual Income")
plt.show()
```
