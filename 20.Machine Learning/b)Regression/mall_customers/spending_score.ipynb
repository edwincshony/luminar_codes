```python
df.drop(columns="CustomerID",inplace=True)
```

```python
X=df.iloc[:,:-1]
```

```python
Y=df.iloc[:,-1]
```

```python
X.head()
```

```python
Y.head()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

X["Genre"]=pd.DataFrame(le.fit_transform(X["Genre"]))
```

```python
Y=pd.DataFrame(le.fit_transform(Y))
```

```python
X.head()
```

```python
Y.head()
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=2)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler=MinMaxScaler()

X_train=pd.DataFrame(scaler.fit_transform(X_train))
X_test=pd.DataFrame(scaler.transform(X_test))
```

```python
from sklearn.neighbors import KNeighborsRegressor

knn=KNeighborsRegressor(n_neighbors=7)

knn.fit(X_train,Y_train)
```

```python
Y_pred_knn=knn.predict(X_test)
```

```python
from sklearn.linear_model import LinearRegression

lr=LinearRegression()

lr.fit(X_train,Y_train)
```

```python
Y_pred_lr=lr.predict(X_test)
```

```python
from sklearn.metrics import r2_score

print("r2_score",r2_score(Y_test,Y_pred_lr)*100)
```

```python
from sklearn.metrics import r2_score

print("r2_score",r2_score(Y_test,Y_pred_knn)*100)
```

```python
from sklearn.tree import DecisionTreeRegressor

dt=DecisionTreeRegressor()

dt.fit(X_train,Y_train)
```

```python
Y_pred_dt=dt.predict(X_test)
```

```python
from sklearn.metrics import r2_score

print("r2_score",r2_score(Y_test,Y_pred_dt))
```

```python
from sklearn.ensemble import AdaBoostRegressor

ada=AdaBoostRegressor()

ada.fit(X_train,Y_train)
```

```python
Y_pred_ada=ada.predict(X_test)
```

```python
from sklearn.metrics import r2_score

print("r2_score",r2_score(Y_test,Y_pred_ada))
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gb=GradientBoostingRegressor()

gb.fit(X_train,Y_train)
```

```python
Y_pred_gb=gb.predict(X_test)
```

```python
from sklearn.metrics import r2_score

print("r2_score",r2_score(Y_test,Y_pred_gb))
```
