```python
import nltk
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
nltk.download('punkt_tab')
sw = stopwords.words("english")
```

```python
message2 = """🙏 We kindly request everyone to attend both sessions regularly without missing any classes,
 as consistent participation is important for your learning and progress. 😊✨"""
```

```python
tokens = nltk.word_tokenize(message2)
print(tokens)
```

```python
refined_tokens = [token for token in tokens if token.lower() not in sw]
refined_tokens
```

```python
import re
new_sentence = " ".join(refined_tokens) # re.sub() works on a string, but refined_tokens is a list.
new_sentence = re.sub("[^A-Za-z0-9 ]","",new_sentence)
new_sentence
```

```python
from nltk import ngrams
NGRAMS = ngrams(nltk.word_tokenize(new_sentence),n=2)
list(NGRAMS)
```
