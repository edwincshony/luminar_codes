### stopwords

```python
from nltk.corpus import stopwords
# nltk.download('stopwords')
# nltk.download('punkt_tab')
sw = stopwords.words("english")
```

```python
print(sw)
```

### word tokenization

```python
sentence = "machine learning is a branch of computer science"
```

```python
tokens = nltk.word_tokenize(sentence)
print(tokens)
```

### stopword removal

```python
refined_tokens = [token for token in tokens if token.lower() not in sw]
```

```python
refined_tokens
```

### special character removal

```python
sentence = "machine learning 🔥🔥🔥🔥🔥🔥 is a branch @ of computer science 🥲"
```

```python
filtered_tokens = []

for w in sentence.split(" "):

    if w.isalnum():

        filtered_tokens.append(w)
print(filtered_tokens)
```

```python
# re=> regular expression => patterns find
```

```python
import re
new_sentence = re.sub("[^A-Za-z0-9 ]","",sentence)
new_sentence
```

```python
from nltk import ngrams
NGRAMS = ngrams(nltk.word_tokenize(new_sentence),n=2)
list(NGRAMS)
```

```python
from nltk.stem import PorterStemmer #SnowballStemmer
ps = PorterStemmer()

words = ["completed","finished","caught","started","preparing","ran"]

for w in words:

    print(f"{w} => {ps.stem(w)}")
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lm = WordNetLemmatizer()

for w in words:

    print(f"{w} => {lm.lemmatize(w,pos="v")}")
```
