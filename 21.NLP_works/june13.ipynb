```python
msg2 = "⚠️ Although the sessions are being attended online, there has been very limited response and participation during the classes. Active engagement is essential to ensure steady progress. 💡📚"
```

```python
import nltk
```

### remove special characters

```python
import re
message = re.sub("[^A-Za-z0-9 ]","",msg2)
message
```

```python
message = message.lower()
message
```

```python
from nltk.corpus import stopwords
# nltk.download('stopwords')
# nltk.download('punkt_tab')
sw = stopwords.words("english")
```

```python
message = [t for t in nltk.word_tokenize(message) if t not in sw]
message = " ".join(message)
message
```

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
stemmed_msg_list = [ps.stem(t) for t in nltk.word_tokenize(message)]
stemmed_msg_list = " ".join(stemmed_msg_list)
stemmed_msg_list
```

```python
from nltk.stem import WordNetLemmatizer
lm = WordNetLemmatizer()
lemmatized_msg_list = [lm.lemmatize(t,pos="v") for t in nltk.word_tokenize(message)]
lemmatized_msg_list = " ".join(lemmatized_msg_list)
lemmatized_msg_list
```
