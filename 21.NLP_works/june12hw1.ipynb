```python
import nltk
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
nltk.download('punkt_tab')
sw = stopwords.words("english")
```

```python
message1 = """📢 From Monday onwards, we will have two sessions every day:
 Deep Learning from 8:30 AM to 10:30 AM and Power BI from 11:00 AM to 1:00 PM. 🚀📚"""
```

```python
tokens = nltk.word_tokenize(message1)
print(tokens)
```

```python
refined_tokens = [token for token in tokens if token.lower() not in sw]
refined_tokens
```

```python
import re
new_sentence = " ".join(refined_tokens) # re.sub() works on a string, but refined_tokens is a list.
new_sentence = re.sub("[^A-Za-z0-9 ]","",new_sentence)
new_sentence
```

```python
from nltk import ngrams
NGRAMS = ngrams(nltk.word_tokenize(new_sentence),n=2)
list(NGRAMS)
```
