### implement positive as 1 and negative as -1

```python
df["sentiment"] = df["sentiment"].map({"positive":1,"negative":-1,"neutral":0})
```

```python
df
```

###convert source to lowercase

```python
df["source"] = df["source"].apply(lambda row:row.lower())
```

```python
df.head()
```

```python
df["review"] = df["review"].apply(lambda row:row.lower())
```

```python
df
```

### remove special characters

```python
import re
df["review"] = df["review"].str.replace("[^A-Za-z0-9 ]","",regex=True)
```

```python
df.head()
```

```python
from nltk.corpus import stopwords
# nltk.download('stopwords')
# nltk.download('punkt_tab')
sw = stopwords.words("english")
```

### remove stopwords from reviews

```python
df["review"] = df["review"].apply(lambda row: [t for t in nltk.word_tokenize(row) if t not in sw]).apply(lambda t: " ".join(t))

```
