### One Hot Encoding

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
categorical_cols = X.select_dtypes(include='object').columns

ct = ColumnTransformer(
    [('encoder', OneHotEncoder(), categorical_cols)],
    remainder='passthrough'
)
X=pd.DataFrame(ct.fit_transform(X))
```

```python
X.dtypes
```

```python
Y.value_counts()
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier()
knn.fit(X_train,Y_train)
Y_predict_knn = knn.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score

print("===== KNN =====")

print("accuracy score",accuracy_score(Y_predict_knn,Y_test)*100)
print("precision_score",precision_score(Y_predict_knn,Y_test)*100)
print("recall_score",recall_score(Y_predict_knn,Y_test)*100)
print("f1_score",f1_score(Y_predict_knn,Y_test)*100)
```

```python
from sklearn.naive_bayes import GaussianNB
nb = GaussianNB()
nb.fit(X_train,Y_train)
Y_predict_nb = nb.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score

print("===== naive bayes =====")


print("accuracy score",accuracy_score(Y_predict_nb,Y_test)*100)
print("precision_score",precision_score(Y_predict_nb,Y_test)*100)
print("recall_score",recall_score(Y_predict_nb,Y_test)*100)
print("f1_score",f1_score(Y_predict_nb,Y_test)*100)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(criterion="entropy",max_depth=4)
dt.fit(X_train,Y_train)
Y_predict_dt = dt.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score

print("===== decision tree =====")

print("accuracy score",accuracy_score(Y_predict_dt,Y_test)*100)
print("precision_score",precision_score(Y_predict_dt,Y_test)*100)
print("recall_score",recall_score(Y_predict_dt,Y_test)*100)
print("f1_score",f1_score(Y_predict_dt,Y_test)*100)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(X_train,Y_train)
Y_predict_rf = rf.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score

print("===== Random Forest =====")

print("accuracy score",accuracy_score(Y_predict_rf,Y_test)*100)
print("precision_score",precision_score(Y_predict_rf,Y_test)*100)
print("recall_score",recall_score(Y_predict_rf,Y_test)*100)
print("f1_score",f1_score(Y_predict_rf,Y_test)*100)
```

```python
from sklearn.ensemble import AdaBoostClassifier
ab = AdaBoostClassifier()
ab.fit(X_train,Y_train)
Y_predict_ada = ab.predict(X_test)
```
